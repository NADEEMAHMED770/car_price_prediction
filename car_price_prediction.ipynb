```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv('car data.csv')
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

```python
df.drop(columns=['Car_Name','Owner'],inplace=True)
```

```python
df['Selling_type'] = df['Selling_type'].map({"Dealer":0,"Individual":1})
```

```python
df['Transmission'] = df['Transmission'].map({"Manual":0,"Automatic":1})
```

```python
fuel_dummies = pd.get_dummies(df['Fuel_Type'], prefix='Fuel', dtype=int)
```

```python
fuel_dummies = fuel_dummies.drop('Fuel_CNG',axis=1)
fuel_dummies.head()
```

```python
df = pd.concat([df, fuel_dummies],axis=1)
```

```python
df.drop('Fuel_Type',axis=1,inplace=True)
```

```python
df.to_csv('car_cleaned_data.csv',index=False)
```

```python
x = df[['Year', 'Present_Price', 'Driven_kms', 'Selling_type',
       'Transmission', 'Fuel_Diesel', 'Fuel_Petrol']]
y = df['Selling_Price']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
results = pd.DataFrame({
    'Actual Price': y_test,
    'Predicted Price': y_pred
})

results.head(10)
```

```python
model.score(x_train,y_train)
```

```python
mae = mean_absolute_error(y_test, y_pred)
print("Mean Absolute Error:", mae)
```

```python
mse = mean_squared_error(y_test, y_pred)
print("Mean Squared Error:", mse)
```

```python
rmse = np.sqrt(mse)
print("Root Mean Squared Error:", rmse)
```

```python
r2 = r2_score(y_test, y_pred)
print("R2 Score:", r2)
```

```python
print("Training Score:", model.score(x_train,y_train))
print("Testing Score:", model.score(x_test,y_test))
```
